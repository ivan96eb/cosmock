# Basic kappa-to-mocks workflow

This notebook is a data-path template. Replace the paths below with local arrays before running the workflow.

Expected shapes:

- `kappa_maps`: `(n_bins, n_pix)`
- `cl_ng`: `(n_bins, n_bins, lmax + 1)`
- `pixwin`: optional `(lmax + 1,)`

If `cl_ng` is not supplied, `cosmock[healpix]` is required so spectra can be estimated from the maps.

```python
import numpy as np
import cosmock as cm

path_to_kappa_maps = "data/kappa_maps.npy"
path_to_cls = "data/target_cls.npy"
path_to_pixwin = "data/pixwin.npy"

nside = 512
order = 3
n_mocks = 10
seed = 1234
```

```python
kappa_maps = np.load(path_to_kappa_maps)
cl_ng = np.load(path_to_cls)
pixwin = np.load(path_to_pixwin)

print("kappa_maps", kappa_maps.shape)
print("cl_ng", cl_ng.shape)
print("pixwin", pixwin.shape)
```

```python
cal = cm.KappaCalibration.from_maps(
    kappa_maps,
    nside=nside,
    cl_ng=cl_ng,
    pixwin=pixwin,
)

model = cm.MockModel.fit(cal, order=order)
```

```python
mocks = model.sample(
    n_mocks=n_mocks,
    seed=seed,
    apply_pixwin=True,
)

print("mocks", mocks.shape)
```

```python
report = model.validate(mocks)

print("finite", report.finite)
print("histogram_l1_by_bin", report.histogram_l1_by_bin)
print("spectra_available", report.spectra_available)
```
